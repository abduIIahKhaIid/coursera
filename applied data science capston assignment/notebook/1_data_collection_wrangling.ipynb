```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

# Web scraping from Wikipedia
url = "https://en.wikipedia.org/wiki/List_of_Falcon_9_and_Falcon_Heavy_launches"
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')
tables = soup.find_all('table', class_='wikitable')
data = pd.read_html(str(tables))[0]

# SpaceX API for launch data
api_url = "https://api.spacexdata.com/v4/launches"
api_data = requests.get(api_url).json()
api_df = pd.DataFrame(api_data)

# Data wrangling
data['Date'] = pd.to_datetime(data['Date'])
data.fillna({'Outcome': 'Unknown'}, inplace=True)
data['Success'] = data['Outcome'].str.contains('Success', case=False).astype(int)

# Save cleaned data
data.to_csv('falcon9_launches.csv', index=False)
api_df.to_csv('spacex_api_data.csv', index=False)
```
