```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
df = pd.read_csv('falcon9_launches.csv')

# Visualization: Success rate by launch site
plt.figure(figsize=(10, 6))
sns.countplot(x='Launch Site', hue='Success', data=df)
plt.title('Launch Success by Site')
plt.savefig('success_by_site.png')
plt.show()

# Visualization: Payload mass vs. success
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Payload Mass (kg)', y='Success', data=df)
plt.title('Payload Mass vs. Landing Success')
plt.savefig('payload_vs_success.png')
plt.show()
```
