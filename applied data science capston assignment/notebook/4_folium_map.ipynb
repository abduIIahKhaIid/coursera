```python
import folium
import pandas as pd

# Load data
df = pd.read_csv('falcon9_launches.csv')

# Create Folium map
m = folium.Map(location=[28.5623, -80.5774], zoom_start=5)  # CCAFS SLC-40
for _, row in df.iterrows():
    folium.Marker(
        location=[row['Latitude'], row['Longitude']],
        popup=f"Site: {row['Launch Site']}, Success: {row['Success']}",
        icon=folium.Icon(color='green' if row['Success'] == 1 else 'red')
    ).add_to(m)

# Save map
m.save('launch_map.html')
```
