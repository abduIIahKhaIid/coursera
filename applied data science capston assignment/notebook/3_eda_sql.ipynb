```python
import sqlite3
import pandas as pd

# Create SQLite database
conn = sqlite3.connect('spacex.db')
df = pd.read_csv('falcon9_launches.csv')
df.to_sql('launches', conn, if_exists='replace', index=False)

# SQL Query: Success rate by launch site
query = """
SELECT "Launch Site", AVG(Success) as Success_Rate
FROM launches
GROUP BY "Launch Site"
"""
result = pd.read_sql(query, conn)
print(result)

# SQL Query: Launches per year
query = """
SELECT strftime('%Y', Date) as Year, COUNT(*) as Launch_Count
FROM launches
GROUP BY Year
"""
result = pd.read_sql(query, conn)
print(result)

conn.close()
```
